**1. Цикл for - базовий**

Використовуючи цикл `for`, виведіть числа від 1 до 7 включно використовуючи функцію range.


```python
for num in range(1, 8):
  print(num)
```

**2. Цикл for по списку**

Дано список:

```python
names = ["Олег", "Анна", "Іван", "Марія"]
```


Використовуючи цикл, виведіть кожне ім'я з привітанням: "Привіт, [ім'я]!"


```python
for name in names:
  print(f"Привіт, {name}!")
```

**3. Цикл з enumerate**

Дано список оцінок:

```python
grades = [85, 92, 78, 95, 88]
```


Використовуючи `enumerate`, виведіть кожну оцінку з номером: "Студент 1: 85 балів"


```python
for i, grade in enumerate(grades, start=1):
  print(f"Студент {i}: {grade} балів")
```

**4. Сума парних чисел**

Напишіть програму, яка обчислює суму всіх парних чисел від 1 до 20 включно.

```python
s = 0
for num in range(1, 21):
  if num % 2 == 0:
    s += num
print(f"Сума всіх парних чисел діапазону [1, 20]: {s}")

print("\n------------------")
print("Якщо треба переглянути вміст:")
print("------------------\n")

even_numbers = []
for num in range(1, 21):
  if num % 2 == 0:
    even_numbers.append(num)
print(f"Всі парні числа діапазону [1, 20] -> {even_numbers}\nЇх сума: {sum(even_numbers)}")
```
